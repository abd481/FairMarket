```python
amenity_map = {

    "Central A/C & heating": "Central AC",

    "Centrally Air-Conditioned": "Central AC",

    "Laundry Facility": "Laundry",

    "Laundry Room": "Laundry",

    "Balcony or Terrace": "Balcony",

    "Covered parking": "Covered Parking"

}
```

```python
amenity_map = {

    "Central A/C & heating": "Central AC",

    "Centrally Air-Conditioned": "Central AC",

    "Laundry Facility": "Laundry",

    "Laundry Room": "Laundry",

    "Balcony or Terrace": "Balcony",

    "Covered parking": "Covered Parking"

}

def normalize_amenities(text):

    if pd.isna(text):
        return None

    items = [
        x.strip()
        for x in text.split(",")
    ]

    normalized = [
        amenity_map.get(item, item)
        for item in items
    ]

    return ", ".join(
        sorted(set(normalized))
    )
```

```python
df_analysis["amenities_clean"] = (
    df_analysis["amenities"]
    .apply(normalize_amenities)
)
```
